```python
import numpy as np
import matplotlib as plt
```

```python
def initialize_board(nrow, mcol, xmine):
    l = np.full((nrow, mcol), '_')

    # Places the mines
    place_mines(l, xmine)

    return l

def place_mines(board, xmine):
    nrow, mcol = len(board), len(board[0])

    # CREATE MINES
    coord_list = [ (x,y) for x in range(nrow) for y in range(mcol) ]    # List of all coords in board
    mines = np.random.choice(len(coord_list), xmine, replace=False)     # choose x random mines in coord_list

    for i in mines:
        x, y = coord_list[i]
        board[y][x] = '*'

def calculate_adjacent_mines(board):
    for i in range(len(board)):
        for j in range(len(board[0])):

            c_count = 0
            for dy in range(-1,2):
                for dx in range(-1,2):
                    x = j + dx
                    y = i + dy

                    if dx == 0 and dy == 0:
                        continue

                    if 0 <= y < len(board) and 0 <= x < len(board[0]):                # Bounds check
                        if board[y][x] == '*':
                            c_count += 1

            if board[i][j] != '*':
                board[i][j] = c_count

def handle_click(board, player_view, row, col):
    if board[row][col] == '0':
        for dx in range(-1,2):
            for dy in range(-1,2):
                x = j + dx
                y = i + dy
                if 0 <= y < len(board) and 0 <= x < len(board[0]):
                    player_view[y][x] = board[y][x]
    else:
        player_view[row][col] = board[row][col]
```

```python
l = initialize_board(5,5,5)
calculate_adjacent_mines(l)

l
```
