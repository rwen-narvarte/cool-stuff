import org.apache.commons.lang3.time.StopWatch;
import org.junit.jupiter.api.Test;

public class Problem15 {
    private static final int SIZE = 20;

    @Test
    public void run() {
        StopWatch watch = new StopWatch();
        watch.start();
        long solution = solve();
        watch.stop();
        System.out.printf("solution: %d took %,.6f ms\n", solution,
        ((double)watch.getNanoTime()) / 1_000_000.0);
    }

    private long solve() {
        return findAllPaths(0, 0);
    }

    private long findAllPaths(int rightCount, int downCount) {
        if (rightCount == downCount) {
            // symmetry => as many paths starting to go right as to go down
            return 2 * findAllPaths(rightCount + 1, downCount );
        }
        if (rightCount >= SIZE || downCount >= SIZE) return 1;
        long count = findAllPaths(rightCount + 1, downCount );
        return count + findAllPaths(rightCount, downCount + 1);
    }
}
