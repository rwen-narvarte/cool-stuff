```python
# Suppose that the triplet a,b,c follows the following constraints:
# a < b < c
# a + b + c = 1000

# it is easy to show that the largest possible value for a can be achieve by minimizing the values of b,c. (Sandwich theorem)
# This value would be at around 1/3 of 1000. More precisely, largest value of a will be 332. 
# Thus, lets set our upper bound lim1 = 332
lim1 = 332  

# On the other hand, we can show that b is within the range of [1000-a < b < 1000 ].
# We can actually make the range narrower by mimimizing and maximizing a and c where [1,2,997] and [1,499,500] are extreme solutions.
# This gives a more solid range for b @ [2,499] and with respect for a, we can say b @ [1000-a,499]
# Thus, lets set our upper bound lim2 = 499
lim2 = 499

for i in range(1, lim1+1):
    for j in range(i, lim2+1):
        a,b = i,j

        # obtaim c using pythagorean theorem
        c = np.sqrt(a**2 + b**2)

        # check if its a valid pythagorean triplet
        if c - int(c) == 0:

            #check if sum == 1000
            if a + b + c == 1000:
                print(a*b*c)
                break
            # else: print(a,b,c)
 
```
