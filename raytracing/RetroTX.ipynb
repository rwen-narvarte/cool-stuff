```python
### IGNORE ###
row = 24
col = 48
canvas = [(i, j) for i in range(row) for j in range(col)]
canvas = np.zeros((row,col))
plt.imshow(canvas)

def background():
    canvas = np.zeros((row,col))
    return canvas
```

```python
root = Tk()                                 # Widget environment
frm = ttk.Frame(root, padding=10)           # Generates window
frm.grid()                                  # Assumes invisible grid on window

# simple ASCII-art for a "blank canvas"
# pre-loaded coordinates for "canvas"
for i,j in canvas:
    ttk.Label(frm, text=f".").grid(row=i, column=j)


# Generate a circle
# r = radius
# C_i = y-component of center
# C_j = x-component of center
def circ(r,C_i,C_j):

    # Compute coordinate of points to be shaded within the circle
    coords = [(i,j) for i in range(row) for j in range(col) if (i-C_i)**2 + (j-C_j)**2 <= r**2]
    for i,j in coords:

        # Shade the coordinates that are found of the list "coords"
        ttk.Label(frm, text=f"%").grid(row=i, column=j)

class circ:
    def __init__(self,r,c_i,c_j):
        self.r = r
        self.c_i = c_i
        self.c_j = c_j
        self.coords = [(i,j) for i in range(row) for j in range(col) if (i-self.c_i)**2 + (j-self.c_j)**2 <= self.r**2]
        
        blankcanvas = np.zeros((row,col))
        for i,j in self.coords: blankcanvas[i][j] = 1
        self.arr = blankcanvas

    def move(self,magnitude,direction):
        if direction == "U":
            self.c_i = self.c_i - magnitude
        if direction == "D":
            self.c_i = self.c_i + magnitude
        if direction == "L":  
            self.c_j = self.c_j - magnitude 
        if direction == "R":
            self.c_j = self.c_j + magnitude
        self.coords = [(i,j) for i in range(row) for j in range(col) if (i-self.c_i)**2 + (j-self.c_j)**2 <= self.r**2]
        
        blankcanvas = np.zeros((row,col))
        for i,j in self.coords: blankcanvas[i][j] = 1
        self.arr = blankcanvas



def render(element):
    frame = background() + element.arr
    plt.imshow(frame)    

# ttk.Button(frm, text="Exit", command=root.destroy).grid(column=2, row=5)
root.mainloop()
```
