```python
from tkinter import *
from tkinter import ttk
import numpy as np
import time
import matplotlib.pyplot as plt
```

```python
### IGNORE ###
row = 5
col = 10
canvas = [(i, j) for i in range(row) for j in range(col)]
canvas = np.zeros((row,col))
plt.imshow(canvas)

def background():
    canvas = np.zeros((row,col))
    return canvas
```

```python
class circ:
    def __init__(self,r,c_i,c_j):
        self.r = r
        self.c_i = c_i
        self.c_j = c_j
        self.coords = [(i,j) for i in range(row) for j in range(col) if (i-self.c_i)**2 + (j-self.c_j)**2 <= self.r**2]
        
        blankcanvas = np.zeros((row,col))
        for i,j in self.coords: blankcanvas[i][j] = 1
        self.arr = blankcanvas

    def move(self,magnitude,direction):
        if direction == "U":
            self.c_i = self.c_i - magnitude
        if direction == "D":
            self.c_i = self.c_i + magnitude
        if direction == "L":  
            self.c_j = self.c_j - magnitude 
        if direction == "R":
            self.c_j = self.c_j + magnitude
        self.coords = [(i,j) for i in range(row) for j in range(col) if (i-self.c_i)**2 + (j-self.c_j)**2 <= self.r**2]
        
        blankcanvas = np.zeros((row,col))
        for i,j in self.coords: blankcanvas[i][j] = 1
        self.arr = blankcanvas

def render(elements):
    frame = background()
    for i in elements:
        frame += i
    plt.imshow(frame)
```

```python
circ2 = circ(5,10,10)
circ2.move(5,"R")
render([circ2.arr])
```
